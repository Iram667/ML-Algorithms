**Problem: Movie Revenue Prediction**

**Import Dependencies**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

**Load the data**

```python
data= "/content/drive/MyDrive/Data sets/revenue-clean.csv"
df=pd.read_csv(data)
df.head()
```

```python
data1= df.copy()
data1.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
x= data1['production_budget_usd']
y= data1['worldwide_gross_usd']
```

**Import matplotlib.pyplot for data plotting**

**plt.scatter()** is a function in matplotlib.pyplot that is used to create scatter plots.


🔹 A **scatter plot** displays individual data points on a graph, typically using dots to show relationships between two variables (X and Y).

```python
plt.scatter(x,y)
```

**Use of Alpha**

🔹 When there are too many points in a scatter plot, they can overlap, making it difficult to see individual points clearly.

🔹 Setting** alpha** (transparency level) helps in better visualizing density by making overlapping points semi-transparent.

```python
plt.scatter(x,y , alpha=0.4)
plt.title('Movie Revenue vs Production Budget')
plt.xlabel('production_budget_usd')
plt.ylabel('worldwide_gross_usd')
plt.show()
```

**Train Algorithm**

**when we want to predict a value based on a relationship between two or more variables, we use linear regression.**

```python
regression = LinearRegression()
print(regression)
```

```python
regression.fit(x.values.reshape(-1,1) , y.values.reshape(-1,1))
#🔹 This line is training a linear regression model using the fit() method from sklearn.linear_model.LinearRegression().
#🔹 It means: "Find the best straight-line relationship between x and y." '''

```

```python
regression.predict(y.values.reshape(-1,1))
```

```python
plt.scatter(x,y , alpha=0.4)
plt.plot(x, regression.predict(x.values.reshape(-1,1)) , color="purple")
plt.title('Movie Revenue vs Production Budget')
plt.xlabel('production_budget_usd')
plt.ylabel('worldwide_gross_usd')
plt.show()
```
